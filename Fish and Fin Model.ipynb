```python
from fastai.vision.all import *
fns = get_image_files('picture Folder')
failed = verify_images(fns)
failed.map(Path.unlink)
len(failed)



dls = ImageDataLoaders.from_folder("picture Folder", valid_pct=0.2, 
                                   item_tfms=Resize(224),bs=2)

learn = vision_learner(dls, resnet18, metrics=error_rate)
learn.fine_tune(3)

is_item, _, prob = learn.predict(PILImage.create('picture Folder/Muna/Muna 1.jpeg'))
print(f'this is an image of {is_item}')
print(f'the probablilty of thiis is {prob[0]:.4f}')


```
